## 1. Walrus Operator `:=`

Ye operator ek value ko assign karta hai **while** condition check ho rahi hai, same line mein.
Normal way mein 2 lines lagti, walrus se 1 hi line mein kaam ho jata hai.


````python
semester_cgpa = [7.8, 8.1, 8.5, 8.6, 8.7]

# yaha len() ka result seedha 'total_sem' variable me store ho raha hai, aur usi line me check bhi ho raha hai
if (total_sem := len(semester_cgpa)) >= 5:
    print(f"Sufyan ne total {total_sem} semesters complete kar liye hain")   # walrus ne ek line bacha di
````

## 2. List Comprehension

Ek chhoti si trick hai naya list banane ke liye, bina for-loop ke 3-4 lines likhe.
Syntax: `[expression for item in list if condition]`


````python
marks = [85, 72, 91, 40, 65]

# har mark pe 5% bonus jodna hai -> normal loop nahi, ek line me hi ho jayega
bonus_marks = [m * 1.05 for m in marks]      # expression: m*1.05, loop: for m in marks
print("Bonus marks:", bonus_marks)

# sirf wahi marks chahiye jo pass hain (>=45) -> condition wala comprehension
passed_marks = [m for m in marks if m >= 45]  # 'if' condition list ke andar hi laga di
print("Passed marks:", passed_marks)
````

## 3. Lambda Functions

Lambda ek chhota, naam-less (anonymous) function hota hai. Jab function bahut simple ho aur baar baar use nahi karna,
tab pura `def` likhne ki jagah ek line ka lambda use kar lete hain.


````python
# normal function hota to 3 lines lagti, lambda se 1 line me ban gaya
add_grace_marks = lambda m: m + 2   # input m, output m+2

print(add_grace_marks(40))   # 42, grace marks mil gaye fail hone se bachne ke liye

# ek aur example: percentage nikalna
calc_percentage = lambda obtained, total: (obtained / total) * 100
print(calc_percentage(450, 500))   # 90.0
````

## 4. map() and filter()

- `map()`: har item pe ek function apply karta hai, poori list pe ek saath
- `filter()`: sirf wahi items rakhta hai jo condition me True aaye

Dono lambda ke saath best kaam karte hain.


````python
marks = [85, 72, 91, 40, 65]

# map() -> har mark me 2 grace marks add karna, sab pe ek saath
with_grace = list(map(lambda m: m + 2, marks))     # map function ko poori list pe chala deta hai
print("With grace marks:", with_grace)

# filter() -> sirf pass wale marks (>=45) chahiye
passed_only = list(filter(lambda m: m >= 45, marks))   # filter sirf True wale items rakhta hai
print("Passed only:", passed_only)
````

## 5. try/except (Basic Error Handling)

Kabhi kabhi code me error aa sakta hai (jaise 0 se divide karna, ya missing data).
`try` block me risky code likhte hain, aur `except` block me batate hain agar error aaye to kya karna hai —
isse program crash nahi hota.


````python
marks = {"Data Structures": 85, "DBMS": 72}

try:
    # 'Maths' subject dictionary me hai hi nahi, isliye KeyError aayega
    print(marks["Maths"])
except KeyError:
    print("Maths ka data abhi available nahi hai, skip kar diya")   # error ko handle kar liya, crash nahi hua

try:
    percentage = 450 / 0   # yaha divide by zero error aayega
except ZeroDivisionError:
    percentage = 0
    print("Total subjects 0 the, isliye percentage 0 rakh diya")

print("Final percentage:", percentage)
````
