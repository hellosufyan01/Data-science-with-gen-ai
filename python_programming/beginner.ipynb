## 1. Variables & Data Types

A variable is a labeled box that stores information.

| Data Type | Meaning | Example |
|---|---|---|
| `str` | Text | `product = "Shoes"` |
| `int` | Whole number | `stock = 50` |
| `float` | Decimal number | `price = 999.99` |
| `bool` | True/False | `in_stock = True` |

Use `type()` to check what kind of data a variable holds.


````python
product = "Shoes"
stock = 50
price = 999.99
in_stock = True

print(product, stock, price, in_stock)
print(type(product), type(stock), type(price), type(in_stock))
````

## 2. Type Casting

Converting one data type into another.
Real use: data from a form or file often comes in as text, even if it's a number.


````python
user_input = "25"          # comes in as text
quantity = int(user_input)  # convert to a real number
total = quantity * 10
print(total)

price_text = "999.50"
price_num = float(price_text)
print(price_num + 50)
````

## 3. Operators

- **Arithmetic**: do calculations (billing)
- **Comparison**: check conditions (is total big enough?)
- **Logical**: combine conditions (is member AND total > 500?)


````python
price = 200
qty = 3
total = price * qty
print("Total:", total)

print(total > 500)          # comparison
print(total == 600)         # equality check

is_member = True
gets_discount = is_member and total > 500   # logical
print("Gets discount:", gets_discount)
````

## 4. Strings

Handling text — like formatting customer names or messages.

Useful string tools: `.title()`, `.upper()`, `.replace()`, slicing `[start:end]`, and f-strings for inserting variables into text.


````python
name = "ravi kumar"

print(name.title())                 # Ravi Kumar
print(name.upper())                 # RAVI KUMAR
print(name.replace("ravi", "raj"))  # raj kumar
print(name[0])                      # first character
print(name[0:4])                    # slice: first 4 characters

print(f"Hello {name.title()}, welcome back!")   # f-string
````

## 5. Conditionals (if / elif / else)

Decision-making in code — like a discount rule engine for a store.


````python
total = 750

if total > 1000:
    discount = 20
elif total > 500:
    discount = 10
else:
    discount = 0

print("Discount %:", discount)
````

## 6. Loops (for / while)

Repeating an action — like processing every order placed in a day.

- `for` loop: repeat for each item in a list
- `while` loop: repeat until a condition becomes False


````python
orders = [250, 800, 1200, 60]

for order in orders:
    print("Order amount:", order)

day = 1
while day <= 3:
    print("Day", day)
    day += 1
````

````python
orders = [
    {"product": "Shoes", "price": 999},
    {"product": "Bag", "price": 499},
    {"product": "Watch", "price": 1999},
    {"product": "Cap", "price": 199},
]
````

````python
for order in orders:
    product = order["product"]
    price = order["price"]

    print(f"{product}: {price}")

    if price > 1000:
        print("High")
    elif price > 500:
        print("Medium")
    else:
        print("Low")

total_float = float(sum(order["price"] for order in orders))
print("Total:", total_float)
````
