## 1. Ternary Operator

A one-line if-else. Useful for quick pass/fail or status checks.


````python
cgpa = 8.5
status = "Pass" if cgpa >= 5 else "Fail"
print("Result status:", status)

attendance = 68
eligible = "Eligible for exam" if attendance >= 75 else "Not eligible, low attendance"
print(eligible)
````

````python
#my prac 
marks= [77,34,27,89,90 ,99]
outcome = ["Pass" if mark >= 35 else "Fail" for mark in marks]
grades = [
    "A GRADE" if mark >= 75
    else "B GRADE" if mark >= 65
    else "C GRADE" if mark >= 35
    else "Fail"
    for mark in marks
]
print("result ",outcome)
print("grades",grades)
````

## 2. Loop Control: break and continue

- `continue`: skip current item, move to next
- `break`: stop the loop completely

Real use: checking subject marks, skip absent subjects, stop once a failing mark is found.


````python
marks = [85, 72, None, 91, 40, 65]   # None means subject not attempted yet

for m in marks:
    if m is None:
        continue          # skip ungraded subject
    if m < 45:
        print("Failing mark found, stopping check:", m)
        break              # stop checking further
    print("Mark:", m)
````

## 3. Unpacking & Swapping

Assign multiple values at once, or swap values without a temp variable.


````python
name, age, roll_no = "Sufyan", 20, 187
print(name, age, roll_no)

sem4_cgpa, sem5_cgpa = 8.6, 8.7
sem4_cgpa, sem5_cgpa = sem5_cgpa, sem4_cgpa   # swap
print("After swap -> sem4:", sem4_cgpa, "sem5:", sem5_cgpa)
````

## 4. enumerate()

Get index and value together while looping. Useful for numbering a report like a semester-wise CGPA sheet.


````python
semester_cgpa = [7.8, 8.1, 8.5, 8.6, 8.7]

for sem_number, cgpa in enumerate(semester_cgpa, start=1):
    print(f"Semester {sem_number}: CGPA {cgpa}")
````

## 5. zip()

Combine two related lists together — like matching subject names with marks.


````python
subjects = ["Data Structures", "DBMS", "Operating Systems", "Computer Networks"]
marks = [85, 72, 91, 65]

for subject, mark in zip(subjects, marks):
    print(f"{subject}: {mark}")
````
