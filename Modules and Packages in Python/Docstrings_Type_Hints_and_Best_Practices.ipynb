## 2. Accessing Docstrings

`help(function_name)` ya `function_name.__doc__` se docstring dekh sakte ho — bade projects me
ye samajhne ke liye kaam aata hai ki koi function kya karta hai bina uska code padhe.


```python
def calculate_percentage(obtained, total):
    """Calculate percentage from obtained marks and total marks."""
    return (obtained / total) * 100

print(calculate_percentage.__doc__)   # sirf docstring text
help(calculate_percentage)             # poori details (docstring + signature)
```

## 3. Type Hints — Basics

Type hints batate hain function ko kis type ka input chahiye aur kis type ka output milega.
Python isse enforce nahi karta (error nahi dega galat type pe), lekin readability aur tools (IDE) ke liye bahut useful hai.


```python
def calculate_percentage(obtained: float, total: float) -> float:
    """Calculate percentage. Takes floats, returns a float."""
    return (obtained / total) * 100

print(calculate_percentage(450, 500))    # normal kaam karta hai
print(calculate_percentage(450, 500.0))  # int/float dono chalta hai, Python enforce nahi karta

def greet_student(name: str) -> str:
    return f"Hello, {name}!"

print(greet_student("Sufyan"))
```

## 4. Type Hints — Complex Types (List, Dict, Optional)

`list[int]`, `dict[str, float]` jaise complex types bhi hint kar sakte ho. `Optional[type]` batata hai
ki value us type ki ho sakti hai ya `None`.


```python
from typing import Optional

def average_marks(marks: list[int]) -> float:
    """Takes a list of ints, returns their average as float."""
    return sum(marks) / len(marks)

def get_student_record(roll_no: int) -> dict[str, float]:
    """Returns a dictionary with name and cgpa."""
    return {"name": "Sufyan", "cgpa": 8.5}

def find_backup_contact(primary: Optional[str] = None) -> str:
    """If no primary contact given, returns a default."""
    if primary is None:
        return "No contact provided"
    return primary

print(average_marks([85, 72, 91, 40]))
print(get_student_record(187))
print(find_backup_contact())
print(find_backup_contact("9999999999"))
```

## 5. Function Naming & Best Practices (PEP 8 Basics)

- Function names **lowercase_with_underscores** hone chahiye (jaise `calculate_percentage`, `CalculatePercentage` nahi)
- Function ka naam bataye ki wo **kya karta hai** (verb se start karo: `get_`, `calculate_`, `is_`, `check_`)
- Ek function **ek hi kaam** kare (agar function bahut lamba ho raha hai, chhote functions me todo)
- Magic numbers avoid karo — seedha `75` likhne ki jagah `MIN_ATTENDANCE = 75` jaisa naam do


```python
# BAD naming (samajh nahi aata kya karta hai, kis type ka data hai)
def calc(x, y):
    return x / y * 100

# GOOD naming (self-explanatory, type hints, docstring)
MIN_ATTENDANCE_REQUIRED = 75    # magic number ki jagah naamdaar constant

def calculate_percentage(obtained: float, total: float) -> float:
    """Return percentage given obtained and total marks."""
    return (obtained / total) * 100

def is_attendance_sufficient(attendance: float) -> bool:
    """Check if attendance meets the minimum requirement."""
    return attendance >= MIN_ATTENDANCE_REQUIRED

print(calculate_percentage(450, 500))
print(is_attendance_sufficient(80))
```
