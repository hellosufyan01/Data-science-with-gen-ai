## 1. Defining & Calling a Function

`def` se function banate hain. Function ka kaam ek baar likho, jitni baar chaho call karo.


```python
def greet_student():
    print("Welcome, Sufyan!")

greet_student()      # call 1
greet_student()      # call 2, same function dobara chala
```

## 2. Parameters & Return Values

Parameters function ko input dete hain. `return` function se output wapas bhejta hai
(jaise `print` sirf dikhata hai, `return` value ko aage use karne ke liye deta hai).


```python
def calculate_percentage(obtained, total):
    percentage = (obtained / total) * 100
    return percentage      # value wapas bheji

result = calculate_percentage(450, 500)   # function call, result store hua
print(result)

# return ke bina bhi call kar sakte ho, lekin value store nahi hogi
def show_percentage(obtained, total):
    print((obtained / total) * 100)   # sirf print, return nahi

show_percentage(450, 500)   # yaha 'result = show_percentage(...)' karoge to None milega
```

## 3. Default Parameter Values

Agar function call karte waqt argument na diya jaye, to default value use hoti hai.


```python
def student_status(name, course="B.Tech"):
    print(f"{name} is enrolled in {course}")

student_status("Sufyan")                  # default course use hoga
student_status("Sufyan", "M.Tech")        # default override ho gaya
```

## 4. Functions Without Return

Agar function me `return` nahi likha, to wo automatically `None` return karta hai.


```python
def log_attendance(status):
    print("Attendance marked:", status)
    # koi return nahi hai

output = log_attendance("Present")
print(output)     # None aayega, kyunki function ne kuch return nahi kiya
```

## 5. Multiple Return Values

Python me ek function ek se zyada values return kar sakta hai — asal me ye ek tuple return karta hai.


```python
def get_student_summary():
    name = "Sufyan"
    roll_no = 187
    cgpa = 8.5
    return name, roll_no, cgpa    # ye asal me ek tuple return kar raha hai: (name, roll_no, cgpa)

# unpacking se seedha 3 variables me le sakte ho
name, roll_no, cgpa = get_student_summary()
print(name, roll_no, cgpa)

# ya poora tuple bhi le sakte ho
summary = get_student_summary()
print(summary, type(summary))
```
