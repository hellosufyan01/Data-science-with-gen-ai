## 1. Positional Arguments

Order matter karta hai — jis order me values do, wahi order me parameters me jaati hain.


```python
def student_info(name, roll_no, course):
    print(f"{name} | Roll: {roll_no} | Course: {course}")

student_info("Sufyan", 187, "B.Tech")     # order sahi
student_info(187, "Sufyan", "B.Tech")     # order galat -> output galat aayega, error nahi
```

## 2. Keyword Arguments

Naam se value do — order matter nahi karta, code zyada readable ban jaata hai.


```python
def student_info(name, roll_no, course):
    print(f"{name} | Roll: {roll_no} | Course: {course}")

student_info(roll_no=187, name="Sufyan", course="B.Tech")   # order kuch bhi ho, sahi jayega
```

## 3. `*args` — Variable Positional Arguments

Jab pata na ho kitne arguments aayenge, `*args` unko ek tuple me collect kar leta hai.


```python
def total_marks(*marks):
    print("Received:", marks, type(marks))   # ye ek tuple hai
    return sum(marks)

print(total_marks(85, 72, 91))          # 3 arguments
print(total_marks(85, 72, 91, 40, 65))  # 5 arguments, koi problem nahi
```

## 4. `**kwargs` — Variable Keyword Arguments

Jab pata na ho kitne named arguments aayenge, `**kwargs` unko ek dictionary me collect karta hai.


```python
def student_profile(**details):
    print("Received:", details, type(details))   # ye ek dictionary hai
    for key, value in details.items():
        print(f"{key}: {value}")

student_profile(name="Sufyan", roll_no=187, course="B.Tech", year=3)
```

## 5. Mixing All Argument Types (Correct Order)

Jab sab types ek sath use karne ho, order fix hai:
`def func(positional, default=value, *args, **kwargs)`


```python
def student_report(name, course="B.Tech", *extra_marks, **details):
    print("Name:", name)
    print("Course:", course)
    print("Extra marks:", extra_marks)     # tuple
    print("Details:", details)             # dictionary

student_report("Sufyan", "B.Tech", 85, 72, 91, roll_no=187, cgpa=8.5)
```
