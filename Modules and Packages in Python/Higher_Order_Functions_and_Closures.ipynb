## 1. Functions as First-Class Objects

Python me function bhi ek normal value ki tarah hai — variable me store kar sakte ho,
list me daal sakte ho, dusre function ko de sakte ho.


```python
def greet_student(name):
    return f"Hello, {name}!"

# function ko variable me store kiya (bina () lagaye, () lagate to call ho jaata)
say_hello = greet_student
print(say_hello("Sufyan"))     # function ki tarah use ho raha hai

# function ko list me store karna
operations = [greet_student, str.upper, str.lower]
print(operations[0]("Sufyan"))   # greet_student call ho raha
```

## 2. Higher-Order Functions (Function as Argument)

Higher-order function wo hota hai jo **kisi aur function ko as argument leta hai**, ya
**function return karta hai**. `map()`, `filter()`, `sorted()` sab higher-order functions hain.


```python
def process_marks(marks_list, func):
    return [func(m) for m in marks_list]

def add_grace(mark):
    return mark + 5

def to_percentage(mark):
    return (mark / 100) * 100

marks = [85, 40, 91, 72]
print(process_marks(marks, add_grace))          # function pass kiya as argument
print(process_marks(marks, to_percentage))      # dusra function pass kiya
```

## 3. Functions Returning Functions

Function ek aur function bhi **return** kar sakta hai — ye advanced patterns ke liye base hai.


```python
def grade_checker(passing_marks):
    def check(marks):                 # inner function
        return "Pass" if marks >= passing_marks else "Fail"
    return check                       # function return ho raha hai, call nahi kiya

strict_check = grade_checker(50)       # passing_marks = 50 wala checker mila
lenient_check = grade_checker(35)      # passing_marks = 35 wala checker mila

print(strict_check(40))     # Fail (50 se kam hai)
print(lenient_check(40))    # Pass (35 se zyada hai)
```

## 4. Closures — Functions that Remember

Closure ek aisa function hai jo apne bahar wale (enclosing) scope ki value ko **yaad rakhta hai**,
chahe outer function khatam ho chuka ho. Upar wala `grade_checker` example bhi ek closure hi tha.


```python
def marks_tracker():
    total = 0                   # ye variable closure me "yaad" rahega

    def add_mark(mark):
        nonlocal total           # outer wale 'total' ko modify karna hai
        total += mark
        return total

    return add_mark

tracker = marks_tracker()        # outer function ek baar chala, ab khatam
print(tracker(85))                # 85, total yaad hai
print(tracker(72))                # 157, purana total add hua
print(tracker(91))                # 248, phir bhi yaad hai
```

## 5. Intro to Decorators (Using Closures)

Decorator ek function hai jo **dusre function ko wrap karke uska behavior badal deta hai**, bina
uske original code ko touch kiye. Ye closures ka hi ek advanced use hai.


```python
def log_call(func):              # decorator function
    def wrapper(*args, **kwargs):     # wrapper -> original function ko call karne se pehle/baad kuch extra karta hai
        print(f"Calling function: {func.__name__}")
        result = func(*args, **kwargs)
        print(f"Finished. Result: {result}")
        return result
    return wrapper

@log_call                          # ye @log_call, calculate_percentage ko log_call() me pass kar deta hai
def calculate_percentage(obtained, total):
    return (obtained / total) * 100

calculate_percentage(450, 500)     # ab call karne pe extra logging bhi ho jayegi
```
