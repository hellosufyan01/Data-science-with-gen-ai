## 1. Lambda Syntax & Basics

`lambda arguments: expression` — ek hi expression, `return` likhne ki zaroorat nahi (khud hi return hoti hai).


```python
square = lambda x: x ** 2
print(square(5))     # 25

is_pass = lambda marks: marks >= 45
print(is_pass(40))   # False
print(is_pass(50))   # True
```

## 2. Lambda with Multiple Arguments

Lambda me multiple arguments comma se separate karke likh sakte ho.


```python
calculate_percentage = lambda obtained, total: (obtained / total) * 100
print(calculate_percentage(450, 500))    # 90.0

add_grace = lambda marks, grace: marks + grace
print(add_grace(40, 5))    # 45
```

## 3. Lambda Inside Other Functions (as arguments)

Lambda ka sabse bada use tab hota hai jab ek function dusre function ko as argument leta hai
(jaise `map`, `filter`, `sorted`) — wahan lambda ek chhota function turant bana ke de deta hai.


```python
def apply_operation(marks_list, operation):
    return [operation(m) for m in marks_list]

marks = [85, 72, 91, 40, 65]

with_bonus = apply_operation(marks, lambda m: m + 5)          # lambda seedha argument ki tarah pass kiya
percentage = apply_operation(marks, lambda m: (m / 100) * 100)

print(with_bonus)
print(percentage)
```

## 4. Lambda with `sorted()` using `key`

`sorted()` ka `key` parameter batata hai kis basis pe sort karna hai — lambda yahan bahut useful hai.


```python
students = [
    {"name": "Sufyan", "cgpa": 8.5},
    {"name": "Raj", "cgpa": 7.9},
    {"name": "Priya", "cgpa": 9.0}
]

sorted_by_cgpa = sorted(students, key=lambda s: s["cgpa"])              # ascending
sorted_by_cgpa_desc = sorted(students, key=lambda s: s["cgpa"], reverse=True)  # descending

print(sorted_by_cgpa)
print(sorted_by_cgpa_desc)

# strings ko length ke hisaab se sort karna
subjects = ["DS", "Operating Systems", "DBMS", "Computer Networks"]
sorted_by_length = sorted(subjects, key=lambda s: len(s))
print(sorted_by_length)
```

## 5. Limitations of Lambda

Lambda sirf **ek expression** tak limited hai — multiple statements, loops, ya multiple lines nahi likh sakte.
Complex logic ke liye normal `def` function hi use karo, lambda sirf chhote, simple kaamo ke liye hai.


```python
# ye kaam lambda se NAHI ho sakta (complex logic, multiple steps):
def grade_calculator(marks):
    if marks >= 90:
        return "A"
    elif marks >= 75:
        return "B"
    elif marks >= 45:
        return "C"
    else:
        return "F"

print(grade_calculator(85))   # B

# lambda sirf simple, one-line kaam ke liye best hai:
simple_check = lambda marks: "Pass" if marks >= 45 else "Fail"
print(simple_check(85))
```
