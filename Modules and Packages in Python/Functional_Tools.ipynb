## 1. `map()` — Apply Function to Every Item

`map(function, iterable)` har item pe function apply karta hai, ek naya map object deta hai
(jise `list()` se list bana sakte ho).


```python
marks = [85, 72, 91, 40, 65]

def add_grace(m):
    return m + 5

with_grace = list(map(add_grace, marks))         # normal function ke sath
with_grace_lambda = list(map(lambda m: m + 5, marks))   # lambda ke sath, same result

print(with_grace)
print(with_grace_lambda)
```

## 2. `filter()` — Keep Only Matching Items

`filter(function, iterable)` sirf wahi items rakhta hai jinke liye function `True` return kare.


```python
marks = [85, 72, 91, 40, 65]

def is_pass(m):
    return m >= 45

passed = list(filter(is_pass, marks))
passed_lambda = list(filter(lambda m: m >= 45, marks))

print(passed)
print(passed_lambda)
```

## 3. `reduce()` — Combine All Items into One Value

`reduce()` `functools` module me hai. Ye poori list ko ek single value me combine kar deta hai,
step by step (jaise sum, product, max nikalna manually).


```python
from functools import reduce

marks = [85, 72, 91, 40, 65]

total = reduce(lambda a, b: a + b, marks)          # sabko add karte jao
print("Total:", total)

highest = reduce(lambda a, b: a if a > b else b, marks)   # sabse bada nikalo
print("Highest:", highest)

# reduce kaise kaam karta hai (step by step):
# step 1: a=85, b=72 -> 85>72 -> 85
# step 2: a=85, b=91 -> 91>85 -> 91
# step 3: a=91, b=40 -> 91
# step 4: a=91, b=65 -> 91
```

## 4. Chaining map, filter, reduce Together

Real-world data processing me in teeno ko ek sath chain karte hain — pehle filter, phir map, phir reduce.


```python
from functools import reduce

marks = [85, 40, 91, 30, 65, 95]

# step 1: sirf pass wale marks (>=45) rakho
passed = list(filter(lambda m: m >= 45, marks))

# step 2: har pass mark pe 5% bonus do
with_bonus = list(map(lambda m: m * 1.05, passed))

# step 3: sabka total nikalo
total = reduce(lambda a, b: a + b, with_bonus)

print("Passed:", passed)
print("With bonus:", with_bonus)
print("Total:", total)
```

## 5. When to Use These vs List Comprehension

List comprehension aksar **zyada readable** hoti hai map/filter se — Python community mostly comprehension
prefer karti hai simple cases ke liye. `map`/`filter`/`reduce` tab useful hain jab function already defined ho,
ya jab functional-programming style chahiye ho (jaise Pandas/PySpark jaisi libraries me ye pattern common hai).


```python
marks = [85, 40, 91, 30, 65]

# map + filter style
result_functional = list(map(lambda m: m + 5, filter(lambda m: m >= 45, marks)))

# same kaam list comprehension se (zyada readable, Pythonic tarika)
result_comprehension = [m + 5 for m in marks if m >= 45]

print(result_functional)
print(result_comprehension)
print(result_functional == result_comprehension)   # dono same result dete hain
```
