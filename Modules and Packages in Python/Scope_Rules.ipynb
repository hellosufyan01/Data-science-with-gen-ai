## 1. Local Scope

Function ke andar banaya gaya variable sirf usi function ke andar accessible hota hai — bahar nahi.


```python
def calculate_cgpa():
    cgpa = 8.5     # ye local variable hai, sirf is function ke andar hi exist karta hai
    print("Inside function:", cgpa)

calculate_cgpa()

try:
    print(cgpa)    # yaha error aayega, cgpa function ke bahar exist hi nahi karta
except NameError as e:
    print("Error:", e)
```

## 2. Global Scope

Function ke bahar banaya gaya variable **global** hota hai — har function ke andar se **read** kiya ja sakta hai
(lekin bina `global` keyword ke modify nahi kar sakte).


```python
student_name = "Sufyan"     # global variable

def greet():
    print("Hello,", student_name)   # global variable ko read kar sakte hain

greet()
print(student_name)    # bahar bhi accessible hai
```

## 3. `global` Keyword

Agar function ke andar se global variable ko **modify** karna ho, `global` keyword use karna padta hai,
warna Python ek naya local variable bana dega usi naam ka.


```python
cgpa = 8.5    # global

def update_cgpa_wrong():
    cgpa = 9.0     # ye galat hai! ye naya LOCAL variable bana deta hai, global change nahi hota
    print("Inside function:", cgpa)

update_cgpa_wrong()
print("Outside (unchanged):", cgpa)   # 8.5 hi rahega

def update_cgpa_correct():
    global cgpa           # ab batate hain ki global cgpa hi use/modify karna hai
    cgpa = 9.0
    print("Inside function:", cgpa)

update_cgpa_correct()
print("Outside (changed):", cgpa)     # ab 9.0 ho jayega
```

## 4. `nonlocal` Keyword (Nested Functions)

Jab function ke andar ek aur function ho (nested), aur andar wala function bahar wale ke variable ko
modify karna chahe, `nonlocal` use karte hain (`global` nahi, kyunki ye outer function ka variable hai, global nahi).


```python
def semester_tracker():
    semester = 1     # outer function ka variable (enclosing scope)

    def move_to_next_semester():
        nonlocal semester       # batate hain ki outer wala 'semester' use karna hai
        semester += 1
        print("Moved to semester:", semester)

    move_to_next_semester()
    move_to_next_semester()
    print("Final semester:", semester)

semester_tracker()
```

## 5. LEGB Rule

Python variable dhoondhte waqt is order me search karta hai:
**L**ocal -> **E**nclosing -> **G**lobal -> **B**uilt-in

Jahan pehle mil jaye, wahi use ho jata hai.


```python
x = "global x"    # Global

def outer():
    x = "enclosing x"   # Enclosing (outer function ka scope)

    def inner():
        x = "local x"    # Local
        print(x)          # sabse pehle Local check hoga -> "local x" milega

    inner()
    print(x)   # yaha "enclosing x" milega (inner ka x sirf uske andar tha)

outer()
print(x)   # yaha "global x" milega

# Built-in example: len, print jaise functions Built-in scope se aate hain
print(len("Sufyan"))   # 'len' kahin define nahi kiya, phir bhi kaam karta hai -> Built-in scope
```
