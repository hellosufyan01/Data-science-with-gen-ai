## 1. What is Recursion — Base Case & Recursive Case

Recursion matlab function khud ko call kare. Har recursive function me 2 cheezein zaroori hain:
- **Base case**: jaha function rukta hai (warna infinite loop ban jayega)
- **Recursive case**: jaha function khud ko chhote problem ke sath call karta hai


```python
def countdown(n):
    if n == 0:                # base case -> yahi pe rukna hai
        print("Liftoff!")
        return
    print(n)
    countdown(n - 1)           # recursive case -> chhote number ke sath khud ko call kiya

countdown(5)
```

## 2. Classic Example: Factorial

`5! = 5 * 4 * 3 * 2 * 1`. Recursion me: `factorial(n) = n * factorial(n-1)`, base case: `factorial(0) = 1`.


```python
def factorial(n):
    if n == 0:                    # base case
        return 1
    return n * factorial(n - 1)   # recursive case

print(factorial(5))    # 5*4*3*2*1 = 120

# semester combinations calculate karne me factorial kaam aata hai (permutations wagera)
print(factorial(4))    # 24
```

## 3. Real-World Style: Summing a List Recursively

List ke total marks nikalne ka recursive tarika — normal `sum()` bhi hai, lekin ye samajhne ke liye
ki recursion data structures pe kaise kaam karta hai.


```python
def sum_marks(marks_list):
    if len(marks_list) == 0:          # base case -> empty list, sum 0 hai
        return 0
    return marks_list[0] + sum_marks(marks_list[1:])   # pehla item + baaki ka recursive sum

marks = [85, 72, 91, 40, 65]
print(sum_marks(marks))    # 353
```

## 4. Recursion with Multiple Branches: Fibonacci

Fibonacci sequence: har number pichle 2 numbers ka sum hota hai (0, 1, 1, 2, 3, 5, 8...).
Isme function khud ko **2 baar** call karta hai — recursion tree banta hai.


```python
def fibonacci(n):
    if n <= 1:                                    # base case
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)     # 2 recursive calls

for i in range(8):
    print(fibonacci(i), end=" ")   # 0 1 1 2 3 5 8 13
print()
```

## 5. Recursion Limits & Why Recursion Isn't Always Best

Python me recursion ki ek default limit hoti hai (usually 1000 calls) — usse zyada gehra jaoge to
`RecursionError` aayega. Fibonacci jaisa recursive function bade `n` ke liye **bahut slow** ho jaata hai
kyunki wahi calculation baar baar repeat hoti hai — normal loop aksar behtar hota hai.


```python
import sys
print("Default recursion limit:", sys.getrecursionlimit())

def infinite_recursion(n):
    return infinite_recursion(n + 1)   # koi base case nahi -> ye kabhi nahi rukega

try:
    infinite_recursion(1)
except RecursionError:
    print("Error: Maximum recursion depth exceeded! (base case missing tha)")

# loop wala factorial -> recursion se zyada fast, memory efficient
def factorial_loop(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result

print(factorial_loop(5))   # 120, same answer, better performance for large n
```
