## 1. Creation & Frozenset

Set `{}` se banta hai, lekin empty set banane ke liye `set()` use karna padta hai
(`{}` se empty **dictionary** ban jaati hai, set nahi).
`frozenset` ek aisa set hai jo immutable hai — change nahi ho sakta.


```python
electives = {"AI", "Cloud Computing", "IoT"}
empty_set = set()          # {} nahi, ye dict ban jaata

# duplicate values automatically hat jaate hain
sem_subjects = {"Maths", "Python", "Maths", "DBMS"}
print(sem_subjects)        # 'Maths' sirf ek baar dikhega

locked_subjects = frozenset(["DS", "DBMS", "OS"])   # immutable set
print(locked_subjects)
```

## 2. Core Methods

`add()` ek item, `remove()`/`discard()` hata dete hain (fark: `remove` na milne pe error deta hai,
`discard` nahi deta), `pop()` random item nikalta hai, `clear()` sab khaali kar deta hai.


```python
electives = {"AI", "Cloud Computing", "IoT"}

electives.add("Blockchain")           # naya item add
electives.discard("Cloud Computing")  # agar na ho to bhi error nahi aayega
print(electives)

try:
    electives.remove("Not There")      # ye error dega kyunki remove strict hai
except KeyError:
    print("Error: item set me tha hi nahi (remove strict hota hai)")

popped_item = electives.pop()          # random item nikaal ke deta hai (order fix nahi)
print("Popped:", popped_item)
print(electives)

electives.clear()
print("After clear:", electives)
```

## 3. Set Operations

Do sets ko compare/combine karne ke operators:
- `|` union: dono ko combine
- `&` intersection: dono me common
- `-` difference: pehle me hai but dusre me nahi
- `^` symmetric_difference: dono me se jo sirf ek me hai (common nahi)

In-place versions bhi hain: `update()`, `intersection_update()`, etc. — ye original set ko modify karte hain.


```python
friend1_electives = {"AI", "Cloud Computing", "IoT"}
friend2_electives = {"IoT", "Blockchain", "AI"}

print("Union:", friend1_electives | friend2_electives)                    # sab combine
print("Intersection:", friend1_electives & friend2_electives)              # dono me common
print("Difference (1-2):", friend1_electives - friend2_electives)          # sirf friend1 ke paas
print("Symmetric Diff:", friend1_electives ^ friend2_electives)            # jo sirf ek ke paas hai

# in-place version -> original set ko hi update kar deta hai
friend1_copy = friend1_electives.copy()
friend1_copy.update(friend2_electives)             # union jaisa, lekin original ko modify karta hai
print("After update:", friend1_copy)

friend1_copy2 = friend1_electives.copy()
friend1_copy2.intersection_update(friend2_electives)   # sirf common rakh do
print("After intersection_update:", friend1_copy2)
```

## 4. Set Comprehension

List comprehension jaisa hi, lekin `{}` use hota hai — output automatically duplicates hata deta hai.


```python
marks = [85, 72, 91, 40, 65, 85, 72]   # duplicate marks bhi hain

unique_marks = {m for m in marks}                 # set comprehension, duplicates gone
pass_marks_set = {m for m in marks if m >= 45}      # conditional set comprehension

print(unique_marks)
print(pass_marks_set)
```

## 5. Use-cases: Deduplication & Fast Membership Testing

Set 2 kaam ke liye best hai:
1. Duplicates hatana (list ko set me convert kar do)
2. `in` check bahut fast hota hai set me, list se zyada tez (bade data pe farak dikhta hai)


```python
# duplicate roll numbers hata na hai (galti se double entry ho gayi thi)
roll_numbers = [187, 145, 187, 200, 145, 187]
unique_rolls = list(set(roll_numbers))     # set banake wapas list me convert
print("Unique rolls:", unique_rolls)

# fast membership check
enrolled_rolls = set(range(100, 300))       # 100 se 299 tak sab enrolled maan lo
print(187 in enrolled_rolls)                # set me check karna list se zyada fast hota hai
```
