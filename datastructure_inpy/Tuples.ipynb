## 1. Creation & Single-element Tuple Gotcha

Tuple `()` se banta hai. Ek common mistake: single-element tuple banate waqt **comma zaroor lagao**,
warna wo tuple nahi, normal value ban jayega.


```python
student_info = ("Sufyan", 187, "B.Tech")   # normal tuple
empty_tuple = ()                            # empty tuple

not_a_tuple = (187)     # ye tuple NAHI hai, ye sirf int hai (bracket sirf grouping ke liye)
real_tuple = (187,)     # comma lagane se ye asli tuple ban gaya

print(type(not_a_tuple))   # <class 'int'>
print(type(real_tuple))    # <class 'tuple'>
```

## 2. Indexing & Slicing

Bilkul list jaisa hi kaam karta hai — indexing, negative indexing, slicing sab possible hai.
Bas ek fark: value change nahi kar sakte.


```python
student_info = ("Sufyan", 20, 187, "B.Tech", 3)

print(student_info[0])       # Sufyan
print(student_info[-1])      # 3 (last item, year)
print(student_info[1:3])     # (20, 187) -> slice
print(len(student_info))     # total items
```

## 3. Tuple Unpacking

Tuple ki values ko seedha alag variables me nikal sakte ho, ek hi line me.
`*rest` se baaki saari values ek list me collect ho jaati hain.


```python
name, age, roll_no, course, year = ("Sufyan", 20, 187, "B.Tech", 3)
print(name, age, roll_no, course, year)

# agar sab values nahi chahiye, sirf pehli aur baaki:
first_sem_marks = (85, 72, 91, 40, 65)
first, *remaining = first_sem_marks
print("First:", first)
print("Remaining:", remaining)   # ye ab ek list hai, tuple nahi

# beech ki values nikalna
first, *middle, last = first_sem_marks
print("First:", first, "Middle:", middle, "Last:", last)
```

## 4. Tuple Methods

Tuple immutable hai isliye iske sirf 2 methods hain — count aur index (list se bahut kam).


```python
marks_tuple = (85, 72, 91, 40, 65, 91)

print(marks_tuple.count(91))    # 91 kitni baar aaya -> 2
print(marks_tuple.index(72))    # 72 ka pehla index kya hai
```

## 5. Immutability — Kab aur Kyun Use Karein

Tuple change nahi ho sakta — isliye jab data **fixed/locked** rahna chahiye, tab tuple use karo.
Common real-world uses: function se multiple values return karna, dictionary keys (list nahi ban sakti key).


```python
# example: function jo ek tuple return karta hai
def get_student_summary():
    return ("Sufyan", 187, 8.5)   # naam, roll, cgpa ek sath return

name, roll, cgpa = get_student_summary()
print(name, roll, cgpa)

# example: tuple ko dictionary key ki tarah use karna (list ye nahi kar sakti)
seat_map = {
    (1, 1): "Sufyan",
    (1, 2): "Raj"
}
print(seat_map[(1, 1)])

try:
    student_info = ("Sufyan", 187)
    student_info[0] = "Raj"   # error aayega, tuple change nahi ho sakta
except TypeError as e:
    print("Error:", e)
```

## 6. Named Tuples (from `collections`)

Normal tuple me `student_info[0]` likhna padta hai, samajh nahi aata ye kya hai.
Named tuple se index ki jagah **naam se access** kar sakte ho — code readable ban jaata hai.


```python
from collections import namedtuple

Student = namedtuple("Student", ["name", "roll_no", "cgpa"])
sufyan = Student(name="Sufyan", roll_no=187, cgpa=8.5)

print(sufyan.name)       # index[0] ki jagah seedha naam se
print(sufyan.roll_no)
print(sufyan.cgpa)
print(sufyan)             # poora record readable format me dikhta hai
```
