```python
original = [1, 2, 3]
alias = original          # ye sirf same list ko point karta hai
alias.append(4)
print(original)           # [1, 2, 3, 4] -> original bhi badal gaya!

safe_copy = original.copy()   # asli copy, alag list
safe_copy.append(99)
print(original)               # ye change nahi hoga
```

```python
squares_1= [x**3 for x in range (10)]
#Step by step:
#range(10) → gives numbers 0, 1, 2, 3, 4

#x**3 → means “square the number” (multiply it by itself)

#0**3 = 0

#1**3 = 1
#2**3=8
#3**3 = 27

#4**3 = 64

#[ ... for x in range(10)] → collect all those results into a list.
print(squares_1)
```

```python
matrix_1=[[i*j for j in range(3)] for i in range(3)]
 #What happens:
#Row 0 → [0*0, 0*1, 0*2] = [0, 0, 0]

#Row 1 → [1*0, 1*1, 1*2] = [0, 1, 2]

#Row 2 → [2*0, 2*1, 2*2] = [0, 2, 4]
print(matrix_1)
```
