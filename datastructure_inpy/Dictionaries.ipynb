## 1. Creation (Multiple Ways)

Dictionary key-value pairs store karta hai. Banane ke 3 tarike: literal `{}`, `dict()` function,
ya `zip()` se do lists ko jodkar.


```python
# tarika 1: literal
sufyan = {"name": "Sufyan", "age": 20, "roll_no": 187}

# tarika 2: dict() function se
sufyan2 = dict(name="Sufyan", age=20, roll_no=187)

# tarika 3: zip() se do lists ko dictionary banana
subjects = ["DS", "DBMS", "OS"]
marks = [85, 72, 91]
marks_dict = dict(zip(subjects, marks))

print(sufyan)
print(sufyan2)
print(marks_dict)
```

## 2. Core Methods

- `.get()`: safely value nikalna (key na ho to error nahi, default deta hai)
- `.keys()`, `.values()`, `.items()`: sab keys/values/pairs nikalna
- `.update()`: multiple values ek sath update
- `.pop()`: specific key hatao aur return karo
- `.popitem()`: last inserted pair hatao
- `.setdefault()`: agar key na ho to default value set karo
- `.clear()`: sab khaali kar do


```python
sufyan = {"name": "Sufyan", "age": 20, "roll_no": 187}

print(sufyan.get("name"))              # normal get
print(sufyan.get("cgpa"))              # key nahi hai -> None (error nahi aata)
print(sufyan.get("cgpa", "Not Set"))   # default value diya

print(list(sufyan.keys()))
print(list(sufyan.values()))
print(list(sufyan.items()))

sufyan.update({"cgpa": 8.5, "year": 3})   # ek sath multiple keys add/update
print(sufyan)

removed_value = sufyan.pop("age")          # key hatao, value return hoti hai
print("Removed age:", removed_value)

last_pair = sufyan.popitem()               # last inserted pair hatao
print("Popped last item:", last_pair)

sufyan.setdefault("branch", "CSE")         # agar 'branch' nahi hai to add karo, hai to kuch mat karo
print(sufyan)
```

## 3. Dictionary Comprehension

`{key: value for item in iterable}` — list comprehension jaisa hi, dictionary ke liye.


```python
subjects = ["DS", "DBMS", "OS", "CN"]
marks = [85, 72, 91, 40]

marks_dict = {s: m for s, m in zip(subjects, marks)}         # basic dict comprehension
passed_only = {s: m for s, m in marks_dict.items() if m >= 45}   # conditional
grade_dict = {s: ("Pass" if m >= 45 else "Fail") for s, m in marks_dict.items()}

print(marks_dict)
print(passed_only)
print(grade_dict)
```

## 4. Nested Dictionaries

Dictionary ke andar dictionary — jaise poora class record, har student apni details ke sath.



```python
class_records = {
    "Sufyan": {"roll_no": 187, "cgpa": 8.5, "year": 3},
    "Raj": {"roll_no": 145, "cgpa": 7.9, "year": 3}
}

print(class_records["Sufyan"])                # poora Sufyan ka record
print(class_records["Sufyan"]["cgpa"])         # sirf uska cgpa

# update nested value
class_records["Sufyan"]["cgpa"] = 8.7
print(class_records["Sufyan"]["cgpa"])

# naya student add karna
class_records["Priya"] = {"roll_no": 200, "cgpa": 9.0, "year": 3}
print(class_records)
```



## 5. Looping Patterns

`.items()` se key aur value dono ek sath mil jaate hain loop me — unpacking se seedha 2 variables me.


```python
marks_dict = {"DS": 85, "DBMS": 72, "OS": 91, "CN": 40}

for subject, mark in marks_dict.items():          # .items() unpacking
    status = "Pass" if mark >= 45 else "Fail"
    print(f"{subject}: {mark} -> {status}")

# sirf keys pe loop
for subject in marks_dict:
    print("Subject:", subject)

# sirf values pe loop
for mark in marks_dict.values():
    print("Mark:", mark)
```

### 🧪 Task 5
`attendance = {"DS": 90, "DBMS": 60, "OS": 85, "CN": 55}`
1. `.items()` loop use karke har subject ke liye print karo "Eligible" (>=75) ya "Not Eligible".
2. Sirf `.values()` pe loop chalake average attendance nikalo.


```python
attendance = {"DS": 90, "DBMS": 60, "OS": 85, "CN": 55}

# 1. Print Eligible/Not Eligible using .items()
for subject, value in attendance.items():
    status = "Eligible" if value >= 75 else "Not Eligible"
    print(f"{subject}: {value} -> {status}")

# 2. Calculate average using .values()
total = sum(attendance.values())
average = total / len(attendance)
print("Average attendance:", average)
```

## 6. defaultdict & Counter (from `collections`)

- `defaultdict`: agar key exist nahi karti, error dene ki jagah automatically default value bana deta hai
- `Counter`: kisi list me har item kitni baar aaya, seedha count kar deta hai


```python
from collections import defaultdict, Counter

# defaultdict: subject-wise marks list banani hai, bina pehle check kiye ki key hai ya nahi
subject_marks = defaultdict(list)
subject_marks["DS"].append(85)
subject_marks["DS"].append(90)     # 2nd internal marks
subject_marks["DBMS"].append(72)
print(dict(subject_marks))

# Counter: kaunsa grade kitni baar aaya
grades = ["A", "B", "A", "C", "B", "A"]
grade_count = Counter(grades)
print(grade_count)
print(grade_count.most_common(1))   # sabse zyada frequent grade
```

### 🧪 Task 6 (Final challenge for Dictionaries)
`attendance_log = ["Present", "Absent", "Present", "Present", "Absent", "Present"]`
1. `Counter` use karke Present aur Absent kitni baar aaye, count karo.
2. `defaultdict(list)` use karke semester-wise CGPA list banao: Semester 1 me 7.8, Semester 1 me 8.0 (revised), Semester 2 me 8.5 — dono values same semester ke andar list me honi chahiye.
3. `most_common()` use karke sabse common attendance status print karo.


```python
from collections import defaultdict, Counter

# 1. Counter for attendance_log
attendance_log = ["Present", "Absent", "Present", "Present", "Absent", "Present"]
attendance_count = Counter(attendance_log)
print("Attendance count:", attendance_count)

# 2. defaultdict for semester-wise CGPA
semester_cgpa = defaultdict(list)
semester_cgpa["Semester 1"].append(7.8)
semester_cgpa["Semester 1"].append(8.0)
semester_cgpa["Semester 2"].append(8.5)
print("Semester CGPA:", dict(semester_cgpa))

# 3. most_common attendance status
most_common_status = attendance_count.most_common(1)
print("Most common status:", most_common_status)
```
