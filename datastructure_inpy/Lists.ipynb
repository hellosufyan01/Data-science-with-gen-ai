## 1. Creation, Indexing, Slicing

List banane ka simple tarika `[]`. Indexing se ek item nikalte hain, slicing se multiple.
Negative index end se count karta hai. Slicing me `step` bhi de sakte ho.


```python
subjects = ["Data Structures", "DBMS", "Operating Systems", "Computer Networks", "Maths"]

print(subjects[0])          # pehla item
print(subjects[-1])         # last item, negative index
print(subjects[1:4])        # index 1 se 3 tak (4 exclude)
print(subjects[::-1])       # poori list reverse (step = -1)
print(subjects[0:5:2])      # step 2 ke sath -> alternate items
print(len(subjects))        # list ki length
print("DBMS" in subjects)   # membership check
```

## 2. Nested Lists (2D data)

List ke andar list — jaise ek marksheet jisme har row me subject aur mark dono hain.


```python
marksheet = [
    ["Data Structures", 85],
    ["DBMS", 72],
    ["Operating Systems", 91],
    ["Computer Networks", 40]
]

print(marksheet[0])        # pehli row -> ['Data Structures', 85]
print(marksheet[0][0])     # subject name
print(marksheet[0][1])     # uska mark

# har subject aur mark print karna
for row in marksheet:
    print(row[0], "->", row[1])
```

## 3. All Core List Methods

Ye sab methods list ko modify karte hain (in-place, koi naya list nahi banta, except copy()).


```python
subjects = ["DS", "DBMS", "OS"]

subjects.append("CN")             # end me ek item add
subjects.insert(1, "Networks")    # specific position (index 1) pe insert
subjects.extend(["AI", "ML"])     # multiple items ek sath add
print(subjects)

subjects.remove("DBMS")           # value dhoond ke remove karo
popped_item = subjects.pop()      # last item remove karo aur return karo
popped_at_index = subjects.pop(0) # specific index se remove
print("Popped:", popped_item, popped_at_index)
print(subjects)

subjects.sort()                   # alphabetically sort, in-place
print("Sorted:", subjects)
subjects.reverse()                # order ulta kar do, in-place
print("Reversed:", subjects)

print(subjects.count("AI"))       # "AI" kitni baar hai
print(subjects.index("CN"))       # "CN" ka index kya hai
subjects.clear()                  # poori list khaali kar do
print("After clear:", subjects)
```

## 4. Copy vs Reference (Common Bug)

Jab tum `list2 = list1` likhte ho, dono **same list** ko point karte hain (alias).
Ek change dusre me bhi dikhega. Real copy ke liye `.copy()` use karo.


```python
original_marks = [85, 72, 91]

alias = original_marks          # ye sirf reference hai, nayi list nahi bani
alias.append(40)
print("Original:", original_marks)   # ye bhi change ho gaya! [85, 72, 91, 40]

safe_copy = original_marks.copy()    # asli independent copy
safe_copy.append(100)
print("Original after safe_copy change:", original_marks)   # ye change nahi hoga
print("Safe copy:", safe_copy)
```

## 5. List Comprehension

`[expression for item in list if condition]` — for-loop ka short form.
Nested comprehension bhi possible hai (list ke andar list banane ke liye).


```python
marks = [85, 72, 91, 40, 65]

squares_of_marks = [m**2 for m in marks]           # basic comprehension
passed = [m for m in marks if m >= 45]              # conditional comprehension
grades = ["Pass" if m >= 45 else "Fail" for m in marks]   # if-else dono ek line me

print(squares_of_marks)
print(passed)
print(grades)

# nested comprehension: 3x3 grid banate hain
grid = [[i*j for j in range(3)] for i in range(3)]
print(grid)
```
